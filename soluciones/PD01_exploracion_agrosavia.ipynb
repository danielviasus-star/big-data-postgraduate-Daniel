## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
#Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.

import pandas as pd
campos = ['Secuencial', 'Fecha de Análisis', 'Departamento', 'Municipio', 'pH agua:suelo']
registros = []

for campo in campos:
    s = ds_agrosavia[campo]
    
    # Nulos nativos de pandas (NaN / None)
    nulos = s.isna().sum()
    
    # Cadenas no nulas con espacios removidos
    s_str = s.dropna().astype(str).str.strip()
    
    # Textos vacíos tras retirar espacios
    textos_vacios = s_str.eq("").sum()
    
    # Marcadores "ND" 
    marcadores_nd = s_str.str.upper().eq("ND").sum()
    
    registros.append({
        "campo": campo,
        "nulos": nulos,
        "textos_vacios": textos_vacios,
        "marcadores_ND": marcadores_nd
    })

# 2. Construir el DataFrame 'perfil' final
perfil = pd.DataFrame(registros)

display(perfil)
````

**Interpretación del resultado:**

Despues de contruir la tabla llamada "perfil", con las columnas Secuencial, Fecha de Análisis, Departamento, Municipio y pH agua:suelo. Se realizo el conteo por separado de los valores nulos, los espacios vacios y los marcadores ND, donde se pudo observar que en la tabla "perfil", solo hay en la columna "Municipios" dos lugares con texto vacio, de resto en la tabla no hay valores nulos, espacios vacios o marcadores ND. Como aclaracion en el punto 1, se le añadio el comando "keep_default_na=False", para que no se tomara automaticamente un espacio vacio o un valor ND, como nulo, porque en caso de que no se colocara se tomaria como valores nulos, los dos espacios vacios que estan en la tabla.

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
# Mostrar los fallos de conversión y sus valores originales.
ds_agrosavia["ph"] = pd.to_numeric(
    ds_agrosavia["pH agua:suelo"].str.strip().str.replace(",", ".", regex=False),
    errors="coerce",
)
ds_agrosavia["fecha_dt"] = pd.to_datetime(
    ds_agrosavia["Fecha de Análisis"],
    format="%d/%m/%Y",
    errors="coerce",
)
#suelo["anio"] = suelo["fecha_dt"].dt.year.astype("Int64")

display(ds_agrosavia[["ph","fecha_dt"]])
````

````python
vacia = ds_agrosavia["pH agua:suelo"].str.strip().eq("")
vacia1 = ds_agrosavia["Fecha de Análisis"].str.strip().eq("")
no_numerica = ~vacia & ds_agrosavia["ph"].isna()
no_numerica1 = ~vacia & ds_agrosavia["fecha_dt"].isna()

print("Vacías_ph:", int(vacia.sum()))
print("Vacías_fecha:", int(vacia1.sum()))
print("No numéricas:", int(no_numerica.sum()))
print(ds_agrosavia.loc[no_numerica, ["Secuencial", "pH agua:suelo"]].head())
print("---------------")
print("No numéricas:", int(no_numerica1.sum()))
print(ds_agrosavia.loc[no_numerica1, ["Secuencial", "Fecha de Análisis"]].head())
````

**Interpretación del resultado:**

Se creo la tabla "suelo" con las columnas "ph" y "fecha_dt", como se vio anteriormente.Tambien se utilizo la conversion numerica y la fecha con formato "%d/%m/%Y". Por otro lado, se observo que no hay vacios ni en la columna 'ph', ni en la columna 'Fecha_dt' pero si se encontro textos no vacios que no se pudo convertir en la columna "ph", por mala redaccion o uso de simbolos como se ve anteriormente. En la columna "fecha_dt" no aparecio texto que no se convirtio.

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
# Comprobar la conservación del total original con assert.
import pandas as pd

en_revision = ds_agrosavia.dropna()
# 1. Identificación individual de cada motivo
motivo_ph_ausente = ds_agrosavia['ph'].isna()
motivo_ph_fuera_rango = ~ds_agrosavia['ph'].between(0, 14) & ds_agrosavia['ph'].notna()
motivo_fecha_invalida = ds_agrosavia['fecha_dt'].isna()  # Suponiendo que las fechas no interpretables son NaT/NaN

# 2. Crear en_revision y los conteos por motivo
en_revision = motivo_ph_ausente | motivo_ph_fuera_rango | motivo_fecha_invalida

# Conteos por motivo separado (usando .loc para cumplir con el apoyo metodológico)
conteo_ph_ausente = len(ds_agrosavia.loc[motivo_ph_ausente])
conteo_ph_fuera_rango = len(ds_agrosavia.loc[motivo_ph_fuera_rango])
conteo_fecha_invalida = len(ds_agrosavia.loc[motivo_fecha_invalida])

# Resumen de motivos
conteos_motivos = pd.Series({
    'pH ausente': conteo_ph_ausente,
    'pH fuera de rango [0, 14]': conteo_ph_fuera_rango,
    'Fecha no interpretable': conteo_fecha_invalida
})

# Distribución general de filas marcadas vs. no marcadas
resumen_revision = en_revision.value_counts()

# 3. Comprobar la conservación del total original con assert
total_original = len(ds_agrosavia)
total_evaluado = resumen_revision.sum()

assert total_evaluado == total_original, (
    f"Error de integridad: La suma de categorías ({total_evaluado}) "
    f"difiere del total de filas original ({total_original})."
)

# Imprimir resultados
display('en_revision')
print("--- Conteos por Motivo ---")
print(conteos_motivos)
print("\n--- Distribución de 'en_revision' ---")
print(resumen_revision)
print(f"\n Validación 'assert' superada: {total_evaluado} filas evaluadas de {total_original} originales.")
````

**Interpretación del resultado:**

Se puede observar que se creo la columna "en_revision", donde se marcaba una fila si su pH es ausente,si su pH esta fuera de [0,14] o su fecha no se pudo interpretar. Pero revisando los resultados, se encontro que existen 27213 filas que cumplen la condicion que el valor de pH esta ausente, mientra que no se presebtaron alguna fila que cumpliera con la condicion de que el pH este fuera de 0 a 14, ni que una fecha no se pueda interpretar. Por otro lado, las filas marcada y las no marcadas, si suma el valor original de filas que tenia pH en ds_agrosavia.
